```python
content = """
Use the conversations below to give a brief summary of an event that occured in the past:
What was the event.
When was the event.
Any other important information.

>>> 2024-02-09 20:47:35.214000+00:00  speedy4134: Looking at the salsa party things got tied
    2024-02-09 20:47:48.196000+00:00  speedy4134: Which does not help in the decision making
    2024-02-09 20:47:56.570000+00:00  speedy4134: Any last minute votes?
    2024-02-09 20:49:02.645000+00:00  schmugungis: I've got dinner plans with my family on Friday, so if it is on Friday, I'll either be really late or can't make it
    2024-02-09 20:49:41.495000+00:00  craizo: I gotta watch harley friday so i cant make it
    2024-02-09 20:49:48.166000+00:00  craizo: If it do be friday
    2024-02-09 20:51:52.467000+00:00  schmugungis: I changed my vote bc originally Friday was the better option but then my sister chucked a wrench into things so now Sunday is the only day fully open
    2024-02-09 20:52:44.448000+00:00  schmugungis: Even then I won't be able to stay super late since I've got midterms

>>> 2024-01-31 02:19:25.592000+00:00  speedy4134: Ok then new event: Salsa party 18th of February. Be there or be a sad tomato

>>> 2024-02-17 16:16:57.169000+00:00  speedy4134: Just a reminder that salsa party is tomorrow! Please bring your favorite/most interesting salsa to share. If you also want to grab a bag of tortilla chips to bring/drink to share/any other party food item you can think of that would be appreciated. But salsa is all that is required. Looking forward to seeing everyone there!

>>> 2024-01-30 22:35:17.323000+00:00  speedy4134: If I hosted a salsa party on Friday the 16 or Sunday the 18th of February would anyone be interested? If so which day?

>>> 2024-01-30 23:13:25.288000+00:00  cacti_entity: stupid question: salsa the dance or salsa the food?
    2024-01-30 23:16:51.978000+00:00  dicoray: both, at once
    2024-01-30 23:17:24.565000+00:00  ledillion: I could be at either
    2024-01-30 23:18:24.023000+00:00  heroshield: def the food but if you wanna teach us Reece👀
    2024-01-30 23:19:21.354000+00:00  schmugungis: I will be frantically studying for the midterm on my birthday, but that would be a good break
    2024-01-30 23:21:28.263000+00:00  craizo: im in for either
    2024-01-30 23:25:29.232000+00:00  speedy4134: Not a stupid question, I am wanting the food, but I'm happy to be taught to salsa
"""
```

```python

def ask(messages: list, model: str = 'qwen3.5:0.8b', think: bool = False) -> dict:
    response = chat(
        model=model,
        messages=messages,
        think=think,
    )
    return {
        'thinking': response.message.thinking or '',
        'content': response.message.content or '',
    }

result = ask([{'role': 'user', 'content': content}])
print(result['content'])
```
